```python
"""
Project Name : LANGHAM Hotel Management System (LHMS)
Author       : Dipson Khatri Chhetri
Student ID   : 850008431
Course       : BBIM502 Introduction to Programming - Assessment 2
Date         : 05/11/2026
Purpose      : Console application that lets hotel staff add and delete rooms,
               view room features, allocate rooms to customers, view current
               allocations, bill a customer and release the room, and save,
               display and back up allocation data using text files.
"""

# ---------------------------------------------------------------------------
# Imports (ImportError is handled so the program exits cleanly if a module
# from the standard library is missing).
# ---------------------------------------------------------------------------
try:
    import ast
    import datetime
    import os
except ImportError as import_error:
    print(f"A required Python module could not be imported: {import_error}")
    raise SystemExit(1)

# ---------------------------------------------------------------------------
# Constants
# ---------------------------------------------------------------------------
STUDENT_ID = "850008431"
GST_RATE = 0.15
LINE = "*" * 70

# Room types: name -> (rate per night, list of features)
ROOM_TYPES = [
    ("Single", 120.00, ["1 single bed", "Free Wi-Fi", "Desk", "City view"]),
    ("Double", 180.00, ["1 queen bed", "Free Wi-Fi", "Mini fridge", "TV"]),
    ("Suite", 350.00, ["1 king bed", "Free Wi-Fi", "Spa bath", "Lounge area", "Lake view"]),
]


# ---------------------------------------------------------------------------
# Classes
# ---------------------------------------------------------------------------
class Room:
    """A hotel room with its type, nightly rate, features and status."""

    def __init__(self, room_no, room_type, rate, features):
        self.room_no = room_no
        self.room_type = room_type
        self.rate = rate
        self.features = features
        self.is_allocated = False


class Customer:
    """A hotel customer."""

    def __init__(self, customer_no, customer_name):
        self.customer_no = customer_no
        self.customer_name = customer_name


class RoomAllocation:
    """Links one customer to one room for a number of nights."""

    def __init__(self, room, customer, nights, allocated_on):
        self.room = room
        self.customer = customer
        self.nights = nights
        self.allocated_on = allocated_on


# ---------------------------------------------------------------------------
# Global application data
# ---------------------------------------------------------------------------
list_of_rooms = []
list_of_allocations = []
activity_log = []


# ---------------------------------------------------------------------------
# File name helpers
# ---------------------------------------------------------------------------
def get_folder_path():
    """Return the folder holding the program so data files are saved beside it."""
    try:
        # __file__ is not defined in some environments (NameError)
        return os.path.dirname(os.path.abspath(__file__))
    except NameError:
        return os.getcwd()


def get_main_file_path():
    """Return the full path of LHMS_<StudentID>.txt."""
    return os.path.join(get_folder_path(), f"LHMS_{STUDENT_ID}.txt")


def get_backup_file_path():
    """Return the full path of LHMS_<StudentID>_Backup_<Date>_<Time>.txt."""
    stamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
    return os.path.join(get_folder_path(), f"LHMS_{STUDENT_ID}_Backup_{stamp}.txt")


def time_text():
    """Return the current date and time as readable text."""
    return datetime.datetime.now().strftime("%d/%m/%Y %H:%M:%S")


def log_activity(message):
    """Record an operation so it can be written to the text file later."""
    activity_log.append(f"[{time_text()}] {message}")


# ---------------------------------------------------------------------------
# Input helpers
# ---------------------------------------------------------------------------
def read_text(prompt):
    """Read a line of text. End of input (EOFError) closes the program safely."""
    try:
        return input(prompt)
    except EOFError:
        print("\nInput ended unexpectedly. Closing the application.")
        raise SystemExit(0)


def read_int(prompt, minimum=None, maximum=None):
    """
    Keep asking until the user enters a whole number inside the range.
    ast.literal_eval is used to turn the text into a Python value, which can
    raise SyntaxError (e.g. '5 +'), ValueError (e.g. 'abc') or, when the value
    is a list or similar, TypeError in int().
    """
    while True:
        text = read_text(prompt)
        try:
            value = ast.literal_eval(text.strip())
            if isinstance(value, (float, str, bool)):
                raise ValueError("not a whole number")
            number = int(value)
            if minimum is not None and number < minimum:
                print(f"Please enter a number of at least {minimum}.")
            elif maximum is not None and number > maximum:
                print(f"Please enter a number no larger than {maximum}.")
            else:
                return number
        except SyntaxError:
            print("That is not valid input. Please enter a whole number.")
        except ValueError:
            print("That is not a whole number. Please try again.")
        except TypeError:
            print("That kind of value is not accepted. Please enter a whole number.")


def read_name(prompt):
    """Keep asking until the user enters a non-empty name."""
    while True:
        name = read_text(prompt).strip()
        if name:
            return name
        print("The name cannot be empty.")


def ask_yes_no(prompt):
    """Return True for y/yes and False for anything else."""
    return read_text(prompt).strip().lower() in ("y", "yes")


def find_room(room_no):
    """Return the Room with this number, or None if it does not exist."""
    for room in list_of_rooms:
        if room.room_no == room_no:
            return room
    return None


def find_allocation(room_no):
    """Return the RoomAllocation for this room number, or None."""
    for allocation in list_of_allocations:
        if allocation.room.room_no == room_no:
            return allocation
    return None


# ---------------------------------------------------------------------------
# Menu option 1 - Add Room
# ---------------------------------------------------------------------------
def choose_room_type():
    """Show the room types and return the chosen (name, rate, features)."""
    print("Room types available:")
    for position, (name, rate, _) in enumerate(ROOM_TYPES, start=1):
        print(f"  {position}. {name} - ${rate:.2f} per night")
    while True:
        choice = read_int("Choose a room type number: ", minimum=1)
        try:
            return ROOM_TYPES[choice - 1]
        except IndexError:
            print(f"There is no room type {choice}. Choose 1-{len(ROOM_TYPES)}.")


def add_rooms():
    """Add one or more new rooms to the list."""
    print("You have selected 'ADD ROOM' from the menu")
    total = read_int("How many rooms would you like to add? ", minimum=1, maximum=100)
    for count in range(1, total + 1):
        print(LINE)
        print(f"Room {count} of {total}")
        room_no = read_int("Enter the room number: ", minimum=1)
        # Keep asking until the number is not already used
        while find_room(room_no) is not None:
            print(f"Room {room_no} already exists.")
            room_no = read_int("Enter a different room number: ", minimum=1)
        name, rate, features = choose_room_type()
        list_of_rooms.append(Room(room_no, name, rate, list(features)))
        log_activity(f"Added room {room_no} ({name}, ${rate:.2f}/night)")
        print(f"Room {room_no} ({name}) has been added.")


# ---------------------------------------------------------------------------
# Menu option 2 - Delete Room
# ---------------------------------------------------------------------------
def delete_rooms():
    """Delete one or more rooms that are not currently allocated."""
    print("You have selected 'DELETE ROOM' from the menu")
    if not list_of_rooms:
        print("There are no rooms to delete. Please add rooms first.")
        return
    keep_going = True
    while keep_going and list_of_rooms:
        room_no = read_int("Enter the room number to delete: ", minimum=1)
        room = find_room(room_no)
        if room is None:
            print(f"Room {room_no} was not found.")
        elif room.is_allocated:
            print(f"Room {room_no} is currently allocated. Bill the customer first.")
        else:
            list_of_rooms.remove(room)
            log_activity(f"Deleted room {room_no}")
            print(f"Room {room_no} has been deleted.")
        keep_going = ask_yes_no("Delete another room? (y/n): ")


# ---------------------------------------------------------------------------
# Menu option 3 - Display Room Details
# ---------------------------------------------------------------------------
def display_room_details():
    """Show the features, rate and status of every room before booking."""
    print("You have selected 'DISPLAY ROOM DETAILS' from the menu")
    if not list_of_rooms:
        print("No rooms to display. Please add rooms first.")
        return
    for room in sorted(list_of_rooms, key=lambda item: item.room_no):
        status = "Occupied" if room.is_allocated else "Available"
        print(LINE)
        print(f"Room Number : {room.room_no}")
        print(f"Room Type   : {room.room_type}")
        print(f"Rate/Night  : ${room.rate:.2f}")
        print(f"Features    : {', '.join(room.features)}")
        print(f"Status      : {status}")


# ---------------------------------------------------------------------------
# Menu option 4 - Allocate Rooms
# ---------------------------------------------------------------------------
def allocate_rooms():
    """Book one or more available rooms for a customer."""
    print("You have selected 'ALLOCATE ROOMS' from the menu")
    available = [room for room in list_of_rooms if not room.is_allocated]
    if not available:
        print("There are no available rooms. Please add rooms or bill a customer first.")
        return
    print(f"There are {len(available)} available room(s).")
    total = read_int("How many rooms would you like to allocate? ",
                     minimum=1, maximum=len(available))
    for count in range(1, total + 1):
        print(LINE)
        print(f"Room Allocation {count} of {total}")
        room = None
        while room is None:
            room_no = read_int("Enter the room number to allocate: ", minimum=1)
            room = find_room(room_no)
            if room is None:
                print(f"Room {room_no} was not found.")
            elif room.is_allocated:
                print(f"Room {room_no} is already occupied.")
                room = None
        customer_no = read_int("Enter the customer number: ", minimum=1)
        customer_name = read_name("Enter the customer name: ")
        nights = read_int("Enter the number of nights: ", minimum=1)
        room.is_allocated = True
        list_of_allocations.append(
            RoomAllocation(room, Customer(customer_no, customer_name), nights, time_text()))
        log_activity(f"Allocated room {room.room_no} to {customer_name} "
                     f"(customer {customer_no}) for {nights} night(s)")
        print(f"Room {room.room_no} has been allocated to {customer_name}.")


# ---------------------------------------------------------------------------
# Menu option 5 - Display Room Allocation Details
# ---------------------------------------------------------------------------
def allocation_text(allocation):
    """Return the lines describing one allocation."""
    return (f"Room Number    : {allocation.room.room_no} ({allocation.room.room_type})\n"
            f"Customer Number: {allocation.customer.customer_no}\n"
            f"Customer Name  : {allocation.customer.customer_name}\n"
            f"Nights Booked  : {allocation.nights}\n"
            f"Allocated On   : {allocation.allocated_on}")


def display_allocations():
    """Show the current room allocation status."""
    print("You have selected 'DISPLAY ROOM ALLOCATION DETAILS' from the menu")
    if not list_of_allocations:
        print("No rooms are currently allocated.")
        return
    for allocation in list_of_allocations:
        print(LINE)
        print(allocation_text(allocation))


# ---------------------------------------------------------------------------
# Menu option 6 - Billing and De-Allocation
# ---------------------------------------------------------------------------
def billing_and_deallocation():
    """Generate a bill for an allocated room and release the room."""
    print("You have selected 'BILLING & DE-ALLOCATION' from the menu")
    if not list_of_allocations:
        print("No rooms are currently allocated, so there is nothing to bill.")
        return
    room_no = read_int("Enter the room number to bill: ", minimum=1)
    allocation = find_allocation(room_no)
    if allocation is None:
        print(f"Room {room_no} is not currently allocated.")
        return
    guests = read_int("How many guests are sharing the bill? (1 = no split): ", minimum=0)
    try:
        # float conversion of a huge night count raises OverflowError
        subtotal = allocation.nights * allocation.room.rate
        gst = subtotal * GST_RATE
        total = subtotal + gst
        # guests = 0 raises ZeroDivisionError
        per_guest = total / guests
    except ZeroDivisionError:
        print("The bill cannot be split between 0 guests. Using 1 guest instead.")
        per_guest = total
    except OverflowError:
        print("That number of nights is too large to bill. The room stays allocated.")
        return
    except TypeError:
        print("The room has no valid rate. The room stays allocated.")
        return

    bill = (f"{LINE}\n"
            f"                 LANGHAM HOTEL - TAX INVOICE\n"
            f"{LINE}\n"
            f"Customer : {allocation.customer.customer_name} "
            f"(No. {allocation.customer.customer_no})\n"
            f"Room     : {allocation.room.room_no} ({allocation.room.room_type})\n"
            f"Nights   : {allocation.nights} x ${allocation.room.rate:.2f}\n"
            f"Subtotal : ${subtotal:,.2f}\n"
            f"GST 15%  : ${gst:,.2f}\n"
            f"TOTAL    : ${total:,.2f}\n"
            f"Per guest: ${per_guest:,.2f}\n"
            f"{LINE}")
    print(bill)
    # Release the room for the next booking
    allocation.room.is_allocated = False
    list_of_allocations.remove(allocation)
    log_activity(f"Billed room {room_no} to {allocation.customer.customer_name}: "
                 f"total ${total:,.2f}. Room released.")
    print(f"Room {room_no} has been released and is available again.")


# ---------------------------------------------------------------------------
# Menu option 7 - Save Room Allocation to a File
# ---------------------------------------------------------------------------
def save_allocations_to_file():
    """Append a snapshot of allocations and activity to LHMS_<StudentID>.txt."""
    print("You have selected 'SAVE ROOM ALLOCATION TO A FILE' from the menu")
    file_path = get_main_file_path()
    try:
        # Mode 'a' creates the file if it does not exist
        data_file = open(file_path, "a")
        try:
            data_file.write(f"{LINE}\nSNAPSHOT SAVED {time_text()}\n{LINE}\n")
            data_file.write("CURRENT ROOM ALLOCATIONS\n")
            if list_of_allocations:
                for allocation in list_of_allocations:
                    data_file.write(allocation_text(allocation) + "\n\n")
            else:
                data_file.write("(none)\n\n")
            data_file.write("ACTIVITY LOG\n")
            if activity_log:
                for entry in activity_log:
                    data_file.write(entry + "\n")
            else:
                data_file.write("(none)\n")
            data_file.write("\n")
        finally:
            data_file.close()
        print(f"Data saved to {os.path.basename(file_path)}")
    except PermissionError:
        print("Permission denied: the file could not be written.")
    except IOError as error:
        print(f"The file could not be written: {error}")


# ---------------------------------------------------------------------------
# Menu option 8 - Display Room Allocation from the File
# ---------------------------------------------------------------------------
def display_allocations_from_file():
    """Read LHMS_<StudentID>.txt and show it on screen."""
    print("You have selected 'DISPLAY ROOM ALLOCATION FROM THE FILE' from the menu")
    file_path = get_main_file_path()
    try:
        data_file = open(file_path, "r")
        try:
            content = data_file.read()
        finally:
            data_file.close()
        if content.strip():
            print(content)
        else:
            print("The file is empty.")
    except FileNotFoundError:
        print("The file does not exist yet. Use option 7 to save data first.")
    except IOError as error:
        print(f"The file could not be read: {error}")


# ---------------------------------------------------------------------------
# Menu option 9 - Backup and Clear Room Allocation File
# ---------------------------------------------------------------------------
def backup_and_clear_file():
    """Append the file's content to a dated backup file, then empty the original."""
    print("You have selected 'BACKUP AND CLEAR ROOM ALLOCATION FILE' from the menu")
    file_path = get_main_file_path()
    backup_path = get_backup_file_path()
    try:
        source_file = open(file_path, "r")
        try:
            content = source_file.read()
        finally:
            source_file.close()
        if not content.strip():
            print("The file is empty, so there is nothing to back up.")
            return
        backup_file = open(backup_path, "a")
        try:
            backup_file.write(content)
        finally:
            backup_file.close()
        # Mode 'w' truncates, which deletes the content of the original file
        clear_file = open(file_path, "w")
        clear_file.close()
        print(f"Content backed up to {os.path.basename(backup_path)}")
        print(f"{os.path.basename(file_path)} has been cleared.")
    except FileNotFoundError:
        print("The file does not exist yet. Use option 7 to save data first.")
    except IOError as error:
        print(f"The backup could not be completed: {error}")


# ---------------------------------------------------------------------------
# Main menu
# ---------------------------------------------------------------------------
MENU_ACTIONS = {
    1: add_rooms,
    2: delete_rooms,
    3: display_room_details,
    4: allocate_rooms,
    5: display_allocations,
    6: billing_and_deallocation,
    7: save_allocations_to_file,
    8: display_allocations_from_file,
    9: backup_and_clear_file,
}


def show_menu():
    """Print the main menu."""
    print(LINE)
    print("               LANGHAM HOTEL MANAGEMENT SYSTEM")
    print(LINE)
    print("1. Add Room")
    print("2. Delete Room")
    print("3. Display Room Details")
    print("4. Allocate Rooms")
    print("5. Display Room Allocation Details")
    print("6. Billing & De-Allocation")
    print("7. Save Room Allocation to a File")
    print("8. Display Room Allocation from the File")
    print("9. Backup and Clear Room Allocation File")
    print("0. Exit Application")
    print(LINE)


def main():
    """Run the menu loop until the user chooses 0."""
    while True:
        show_menu()
        choice = read_int("Enter your choice (0-9): ")
        if choice == 0:
            print("Thank you for using the LANGHAM Hotel Management System. Goodbye!")
            break
        try:
            action = MENU_ACTIONS[choice]
        except KeyError:
            print("Invalid choice. Please enter a number from 0 to 9.")
            print()
            continue
        try:
            action()
        except NameError as error:
            print(f"An internal name error occurred: {error}")
        print()


if __name__ == "__main__":
    main()
```
